## Loading the Dataset:

```python
import pandas as pd

df = pd.read_csv(r'/content/Global_Superstore2.csv', encoding='latin1')
```

## Exploring the Dataset

We examine the first few records and inspect the structure of the dataset using `head()` and `info()`.

```python
df.head() #view data
df.info() #dataset information
```

## Identifying Missing Values

We check each column for missing values. The `Postal Code` column contains a large number of missing values.
## Handling Missing Values

The `Postal Code` column contained a large number of missing values (over 80% of the entries). Since the missing data was too extensive to impute reliably, the column was removed from the dataset.

```python
print(df.isnull().sum()) # Check Missing Values
```

```python
df = df.drop(columns=["Postal Code"])
```

## Checking for Duplicate Records

We check for duplicate rows and remove them if any are found.

```python
duplicates = df.duplicated().sum() ## Check Duplicates
print("Duplicate Rows:", duplicates)
```

```python
print(df.columns)

print(df.shape)

df.info()
```

## Converting Data Types

The `Order Date` and `Ship Date` columns are converted from string (`object`) to `datetime`.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"]) ## Convert Date Columns
df["Ship Date"] = pd.to_datetime(df["Ship Date"])  # Convert Date Columns
df.info() ## Check Updated Data Types
```

## Exporting the Cleaned Dataset

The cleaned dataset is saved as a new CSV file.

```python
df.to_csv("Global_Superstore_Cleaned.csv", index=False)
```
